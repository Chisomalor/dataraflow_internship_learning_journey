```python
num = int(input("Enter a number: "))

for i in range(1, 13):
    result = num * i
    print(f"{num} * {i} = {result}")
```

# Task 5-Calculate the factorial of a number using a loop

```python
number = int(input("Enter a number: "))

factorial = 1

for i in range(1, number + 1):
    factorial *= i

    print(f"The factorial of {number} is {factorial}")
```

# Task 7-Count vowels in a string

# Task 8-Function that Returns sum, difference, and product of two numbers

```python
def calculate(a, b):
    total_sum = a + b
    difference = a - b
    product = a * b
    return total_sum, difference, product

# Calling the function and unpacking the returned values
s, d, p = calculate(15, 5)

print(f"Sum: {s}")
print(f"Difference: {d}")
print(f"Product: {p}")
```

# Task 9-Reverse a string WITHOUT using slicing ([::-1])

```python
original_text = input("Enter a word: ")
reversed_text = ""

for letter in original_text:
    reversed_text = letter + reversed_text

print(f"Original: {original_text}")
print(f"Reversed: {reversed_text}")
```

# Task 10-First 10 numbers of the Fibonacci sequence

```python
# Starting values for the sequence
a, b = 0, 1

fibonacci_sequence = []

for _ in range(10):
    fibonacci_sequence.append(a)
    a, b = b, a + b

    print(f"The first 10 Fibonacci numbers are: {fibonacci_sequence}")
```

# Assignment 1-Student grade system

```python
total_marks = 0

for i in range(1, 6):
    mark = float(input(f"Enter mark for subject {i}: "))
    total_marks += mark

    average = total_marks / 5

    if average >= 80:
        grade = "A"
    elif average >= 60:
        grade = "B"
    elif average >= 40:
        grade = "C"
    else:
        grade = "F"

    print(f"\nTotal Marks: {total_marks} / 500")
    print(f"Average Mark: {average:.2f}")
    print(f"Final Grade: {grade}")
```

# Assignment 2-Simple calculator

```python
# 1. Show menu
print("Select operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

# 2. Collect user input
choice = input("Enter choice (1/2/3/4): ")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# 3. Perform the calculation based on choice
if choice == "1":
   result = num1 + num2
   print(num1, "+", num2, "=", result)

elif choice == "2":
     result = num1 - num2
     print(num1, "-", num2, "=", result)

elif choice == "3":
     result = num1 * num2
     print(num1, "*", num2, "=", result)

elif choice == "4":
    if num2 == 0:
        print("Error! Cannot divide by zero.")
    else:
        result = num1 / num2
        print(num1, "/", num2, "=", result)

else:
    print("Invalid choice!")
```

# Assignment 3-Shopping cart

```python
# Create an empty dictionary to hold items and prices
cart = {}

while True:
     item = input("Enter item name (or type 'done' to finish): ")

     if item.lower() == "done":
         break # Stop adding items

     price = float(input("Enter price for " + item + ": "))

     # Store item as key and price as value in the dictionary
     cart[item] = price

# Calculate total cost using dictionary values
total_cost = sum(cart.values())

# Display total cost
print("\nItems in your cart:", cart)
print("Total Cost: $", total_cost)
```

# Assignment 4-Guessing game

```python
import random

# Computer picks a secret number between 1 and 20
secret_number = random.randint(1, 20)

print("Welcome to the Guessing Game!")
print("I'm thinking of a number between 1 and 20.")

# Loop keeps running until the user guesses correctly
while True:
    guess = int(input("Take a guess: "))

    if guess > secret_number:
       print("Too high! Try again.")
    elif guess < secret_number:
       print("Too low! Try again.")
    else:
       print("Congratulations! You guessed the secret number!")
       break # Game over, exit the loop
```

# Assignment 5-File handling

```python
# STEP 1: Collect student names from user and write them to a file
with open("students.txt", "w") as file:
     print("Enter student names (type 'done' when finished):")

     while True:
          name = input("Enter student name: ")

          if name.lower() == "done":
             break # Stop taking names

          # Write the name to the file with a new line (\n)
          file.write(name + "\n")

print("\nNames successfully saved to 'students.txt'!")

# STEP 2: Open the file in read mode and display the names
print("\n--- Reading Student Names From File ---")

with open("students.txt", "r") as file:
     for line in file:
          # .strip() removes the extra new line space at the end of each line
         print(line.strip())
```

# Assessment 6-Local vs Global variables

Global variable is defined outside any function, can be seen and used anywhere in your code.

Local variable is defined inside a function, only exists and can be used within that function; it appears once the function finishes running.

Below is an example:



```python
total_seeds = 100 # global variable

def show_germinated():
    germinated = 60 # local variable, only exists inside this function
    print(germinated, "out of", total_seeds)

show_germinated()
```

if we try to print(germinated) outside the function, Python will throw a NameError, because germinated only exists inside show_germinated() and does not exist anywhere else in the code.

# Assessment 7-Function that returns only even numbers from a list


```python
# Function to filter even numbers
def get_even_numbers(numbers_list):
    even_list = [] # Empty list to hold our even numbers

    for num in numbers_list:
        if num % 2 == 0: # Check if number is even
            even_list.append(num)

    return even_list

# Test data (e.g., harvested crop weights in kg)
sample_numbers = [12, 7, 15, 20, 33, 42, 50]

# Call the function and store the returned list
result = get_even_numbers(sample_numbers)

print("Original List:", sample_numbers)
print("Even Numbers Only:", result)
```
